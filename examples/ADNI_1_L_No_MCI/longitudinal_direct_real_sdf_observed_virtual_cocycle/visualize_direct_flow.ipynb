# Direct-flow longitudinal volume and change visualization

This notebook is organized around:
- split summaries and dashboards;
- one representative subject-case from `train`, `val`, and `test`;
- real-age volume comparison against `direct`, `composed`, and `no-change` forecasts;
- out-of-distribution forecasts to ages `92`, `95`, `100`, and `105`;
- rigid-aligned surface-shift heatmaps instead of transparent mesh-overlap plots;
- anchor-age `60`, `70`, and `80` twenty-year forecasts for both `CN` and `AD`.

Every figure is shown inline and also saved as a separate HTML file under `analysis/notebook_<checkpoint>/` inside the selected experiment.

````python
from pathlib import Path
import sys
from IPython.display import Markdown, display

def find_repo_root(start: Path) -> Path:
    current = start.resolve()
    for candidate in [current, *current.parents]:
        if (candidate / '.git').exists() or (candidate / 'deep_sdf').exists():
            return candidate
    raise RuntimeError('Could not find repository root.')

REPO = find_repo_root(Path.cwd())
HELPER_DIR = REPO / 'examples' / 'ADNI_1_L_No_MCI'
for extra in (REPO, HELPER_DIR):
    if str(extra) not in sys.path:
        sys.path.insert(0, str(extra))

EXPERIMENT = (REPO / 'examples' / 'ADNI_1_L_No_MCI' / 'longitudinal_direct_full256_siren_sequence_cocycle_latent_guard').resolve()
CHECKPOINT = 'best'
DEVICE = 'auto'

CLEAR_PREVIOUS_HTML = True

MAIN_MESH_RESOLUTION = 96
ANCHOR_MESH_RESOLUTION = 80
MESH_MAX_BATCH = 2 ** 18
OOD_AGES = [92.0, 95.0, 100.0, 105.0]
COMPOSED_STEP_YEARS = 0.5
ANCHOR_AGES = [60.0, 70.0, 80.0]
ANCHOR_HORIZON_YEARS = 20.0
ANCHOR_EVALUATION_STEP_YEARS = 2.0
CHANGE_SAMPLE_COUNT = 3000
BATCH_TREND_TRAIN_SUBJECT_LIMIT = 50
BATCH_TREND_MESH_RESOLUTION = 80

EXPERIMENT
````

````python
import direct_flow_rich_notebook_helpers as helpers

bundle = helpers.load_bundle(EXPERIMENT, checkpoint=CHECKPOINT, device=DEVICE)
removed_html = helpers.clear_generated_html(bundle) if CLEAR_PREVIOUS_HTML else []

display(Markdown(f'### Experiment\n`{bundle.experiment_dir}`'))
display(Markdown(f'### Checkpoint\n`{bundle.checkpoint_name}` at epoch `{bundle.checkpoint["epoch"]}`'))
display(Markdown(f'### Device\n`{bundle.device}`'))
display(Markdown(f'### HTML output directory\n`{bundle.html_dir}`'))
if removed_html:
    display(Markdown(f'### Removed previous notebook HTML files\nCount: `{len(removed_html)}`'))
bundle.html_dir
````

````python
available_splits = helpers.available_summary_splits(bundle)
display(Markdown('### Available evaluated splits'))
display(available_splits)
if 'train' not in available_splits:
    display(Markdown(
        '### Missing train summary\n'
        'Run this once if you want train/val/test summaries from the same checkpoint:\n\n'
        '```bash\n'
        f'/home/jakaria/anaconda3/envs/inr_sdf/bin/python evaluate_deep_sdf_longitudinal_direct_flow.py -e {bundle.experiment_dir} --checkpoint {bundle.checkpoint_name} --split all --gpu 0\n'
        '```'
    ))

display(Markdown('### Overall split summary'))
display(helpers.summary_table(bundle).round(6))

sequence_splits = helpers.available_sequence_summary_splits(bundle)
if sequence_splits:
    display(Markdown('### Sequence summaries available'))
    display(sequence_splits)

for split in available_splits:
    display(Markdown(f'### {split.upper()} diagnosis summary'))
    display(helpers.diagnosis_summary_table(bundle, split).round(6))
    display(Markdown(f'### {split.upper()} pair-type summary'))
    display(helpers.pair_type_summary_table(bundle, split).round(6))
    if split in sequence_splits:
        display(Markdown(f'### {split.upper()} sequence diagnosis summary'))
        display(helpers.sequence_diagnosis_summary_table(bundle, split).round(6))
````

````python
dashboard_figures = helpers.build_all_dashboard_figures(bundle)
dashboard_paths = {}
for split, fig in dashboard_figures.items():
    html_path = helpers.save_figure(bundle, fig, f'{split}_dashboard')
    dashboard_paths[split] = html_path
    display(Markdown(f'### {split.upper()} dashboard\nSaved to `{html_path}`'))
    fig.show()
dashboard_paths
````

````python
batch_trend_dataset = helpers.build_observed_age_volume_trend_dataset(
    bundle,
    splits=('train', 'val', 'test'),
    subject_limits={
        'train': BATCH_TREND_TRAIN_SUBJECT_LIMIT,
        'val': None,
        'test': None,
    },
    mesh_resolution=BATCH_TREND_MESH_RESOLUTION,
    mesh_max_batch=MESH_MAX_BATCH,
)

trend_frame = batch_trend_dataset['trend_frame']
subject_summary = batch_trend_dataset['subject_summary']
decode_status = batch_trend_dataset['decode_status']

trend_csv = bundle.html_dir / 'batch_observed_age_volume_trend.csv'
subject_csv = bundle.html_dir / 'batch_observed_age_subject_summary.csv'
decode_csv = bundle.html_dir / 'batch_observed_age_decode_status.csv'
trend_frame.to_csv(trend_csv, index=False)
subject_summary.to_csv(subject_csv, index=False)
decode_status.to_csv(decode_csv, index=False)

display(Markdown('## Batch observed-age volume trend'))
display(Markdown(
    f'Train uses `{BATCH_TREND_TRAIN_SUBJECT_LIMIT}` subjects. '
    'Validation and test use all available subjects. '
    'Predictions are decoded at the exact observed scan ages for each subject.'
))
display(helpers.observed_age_volume_summary_table(batch_trend_dataset).round(6))

age_fig = helpers.observed_age_volume_trend_figure(trend_frame)
age_path = helpers.save_figure(bundle, age_fig, 'batch_observed_age_volume_trend')
display(Markdown(f'### Absolute volume vs real age\nSaved to `{age_path}`'))
age_fig.show()

elapsed_fig = helpers.elapsed_relative_volume_trend_figure(trend_frame)
elapsed_path = helpers.save_figure(bundle, elapsed_fig, 'batch_elapsed_relative_volume_trend')
display(Markdown(f'### Relative volume vs elapsed years from baseline\nSaved to `{elapsed_path}`'))
elapsed_fig.show()

decode_failures = decode_status.loc[~decode_status['decode_success'].astype(bool)].copy()
display(Markdown('### Batch decode failures'))
if decode_failures.empty:
    display(Markdown('No decode failures in this observed-age batch.'))
else:
    display(decode_failures.round(6))

display(Markdown('### Saved batch trend files'))
batch_trend_outputs = {
    'absolute_age_volume_html': age_path,
    'elapsed_relative_volume_html': elapsed_path,
    'trend_csv': trend_csv,
    'subject_summary_csv': subject_csv,
    'decode_status_csv': decode_csv,
}
batch_trend_outputs
````

````python
split_cases = {}
split_case_html = {}

for split in ('train', 'val', 'test'):
    if split not in available_splits:
        continue
    pair_row = helpers.select_representative_pair(
        bundle,
        split=split,
        pair_type='nonadjacent',
        require_observed_intermediate=True,
        max_source_age_years=min(OOD_AGES),
    )
    case = helpers.build_subject_forecast_case(
        bundle,
        pair_row,
        ood_ages_years=OOD_AGES,
        composed_step_years=COMPOSED_STEP_YEARS,
        mesh_resolution=MAIN_MESH_RESOLUTION,
        mesh_max_batch=MESH_MAX_BATCH,
    )
    split_cases[split] = case

    display(Markdown(f'## {split.upper()} representative subject case'))
    display(helpers.subject_case_overview_table(case).round(6))
    display(Markdown(f'### {split.upper()} decode status'))
    display(helpers.subject_decode_status_table(case).round(6))

    volume_fig = helpers.subject_forecast_volume_figure(case)
    volume_path = helpers.save_figure(bundle, volume_fig, f'{split}_subject_volume_forecast')
    display(Markdown(f'### {split.upper()} volume forecast\nSaved to `{volume_path}`'))
    volume_fig.show()

    display(Markdown(f'### {split.upper()} observed-age volume comparison'))
    display(helpers.subject_observed_volume_error_table(case).round(6))

    observed_fig, observed_summary = helpers.observed_target_change_heatmap_figure(
        case,
        sample_count=CHANGE_SAMPLE_COUNT,
    )
    observed_path = helpers.save_figure(bundle, observed_fig, f'{split}_subject_observed_target_change')
    display(Markdown(f'### {split.upper()} observed-target change map\nSaved to `{observed_path}`'))
    observed_fig.show()
    display(Markdown(f'### {split.upper()} observed-target change summary'))
    display(observed_summary.round(6))

    split_case_html[split] = {
        'volume_forecast': volume_path,
        'observed_target_change': observed_path,
    }

    if case['ood_ages_years']:
        direct_ood_fig, direct_ood_summary = helpers.ood_change_heatmap_figure(
            case,
            method='direct',
            sample_count=CHANGE_SAMPLE_COUNT,
        )
        direct_ood_path = helpers.save_figure(bundle, direct_ood_fig, f'{split}_subject_ood_direct_change')
        display(Markdown(f'### {split.upper()} OOD direct change maps\nSaved to `{direct_ood_path}`'))
        direct_ood_fig.show()
        display(Markdown(f'### {split.upper()} OOD direct change summary'))
        display(direct_ood_summary.round(6))

        composed_ood_fig, composed_ood_summary = helpers.ood_change_heatmap_figure(
            case,
            method='composed',
            sample_count=CHANGE_SAMPLE_COUNT,
        )
        composed_ood_path = helpers.save_figure(bundle, composed_ood_fig, f'{split}_subject_ood_composed_change')
        display(Markdown(f'### {split.upper()} OOD composed change maps\nSaved to `{composed_ood_path}`'))
        composed_ood_fig.show()
        display(Markdown(f'### {split.upper()} OOD composed change summary'))
        display(composed_ood_summary.round(6))

        split_case_html[split]['ood_direct_change'] = direct_ood_path
        split_case_html[split]['ood_composed_change'] = composed_ood_path

split_case_html
````

````python
anchor_cases = {}
anchor_html = {}

for diagnosis in ('CN', 'AD'):
    for anchor_age in ANCHOR_AGES:
        case = helpers.build_anchor_age_case(
            bundle,
            anchor_age_years=anchor_age,
            diagnosis=diagnosis,
            horizon_years=ANCHOR_HORIZON_YEARS,
            evaluation_step_years=ANCHOR_EVALUATION_STEP_YEARS,
            composed_step_years=COMPOSED_STEP_YEARS,
            mesh_resolution=ANCHOR_MESH_RESOLUTION,
            mesh_max_batch=MESH_MAX_BATCH,
        )
        key = f'{diagnosis}_{int(anchor_age)}'
        anchor_cases[key] = case

        display(Markdown(f'## {diagnosis} anchor age {anchor_age:.0f}'))
        display(helpers.anchor_case_overview_table(case).round(6))
        display(Markdown(f'### {diagnosis} anchor {anchor_age:.0f} decode status'))
        display(helpers.anchor_decode_status_table(case).round(6))

        volume_fig = helpers.anchor_volume_figure(case)
        volume_path = helpers.save_figure(bundle, volume_fig, f'anchor_{diagnosis.lower()}_{int(anchor_age)}_volume')
        display(Markdown(f'### {diagnosis} anchor {anchor_age:.0f} volume forecast\nSaved to `{volume_path}`'))
        volume_fig.show()

        change_fig, change_summary = helpers.anchor_change_heatmap_figure(
            case,
            sample_count=CHANGE_SAMPLE_COUNT,
        )
        change_path = helpers.save_figure(bundle, change_fig, f'anchor_{diagnosis.lower()}_{int(anchor_age)}_change')
        display(Markdown(f'### {diagnosis} anchor {anchor_age:.0f} 20-year change map\nSaved to `{change_path}`'))
        change_fig.show()
        display(Markdown(f'### {diagnosis} anchor {anchor_age:.0f} change summary'))
        display(change_summary.round(6))

        anchor_html[key] = {
            'volume_forecast': volume_path,
            'change_map': change_path,
        }

anchor_html
````
