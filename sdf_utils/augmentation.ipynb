```python
import os
import trimesh
import numpy as np
```

```python
source_dir = "../../../torus_bump_5000_two_scale_binary_bump_variable_noise_fixed_angle/scaled_obj_files"
output_dir = "../../../torus_bump_5000_two_scale_binary_bump_variable_noise_fixed_angle/scaled_obj_files_aug"
```

```python
def random_rotation_matrix(max_angle_degrees):
    angle = np.radians(np.random.uniform(5, max_angle_degrees))
    axis = np.random.uniform(-1, 1, 3)
    axis /= np.linalg.norm(axis)
    return trimesh.transformations.rotation_matrix(angle, axis)

def apply_random_transformations(mesh, max_angle_degrees):
    rotation_matrix = random_rotation_matrix(max_angle_degrees)
    mesh.apply_transform(rotation_matrix)
    return mesh
```

```python
os.makedirs(output_dir, exist_ok=True)

for filename in os.listdir(source_dir):
    if filename.endswith(".obj"):
        mesh = trimesh.load(os.path.join(source_dir, filename))
        
        for i in range(2):
            transformed_mesh = apply_random_transformations(mesh.copy(), 15)
            output_filename = f"{os.path.splitext(filename)[0]}_transformed_{i}.obj"
            transformed_mesh.export(os.path.join(output_dir, output_filename))
```
