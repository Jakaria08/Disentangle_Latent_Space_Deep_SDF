```python
import os
import trimesh
import numpy as np
```

```python
source_dir = "../../../final_classification_dataset_femur/all_mesh/obj_files"
output_dir = "../../../final_classification_dataset_femur/all_mesh/obj_files_aug"
```

```python
import numpy as np
import trimesh

def random_rotation_matrix(max_angle_degrees):
    # Sample within ±(max_angle_degrees/2) to keep total range as max_angle_degrees
    angle = np.radians(np.random.uniform(-max_angle_degrees/2, max_angle_degrees/2))
    axis = np.random.uniform(-1, 1, 3)
    axis /= np.linalg.norm(axis)
    return trimesh.transformations.rotation_matrix(angle, axis)

def random_scaling_matrix(min_scale, max_scale):
    scale = np.random.uniform(min_scale, max_scale)
    return trimesh.transformations.scale_matrix(scale)

def random_translation_matrix(max_translation):
    translation = np.random.uniform(-max_translation, max_translation, 3)
    return trimesh.transformations.translation_matrix(translation)

def apply_random_transformations(mesh, max_angle_degrees=10, min_scale=0.95, max_scale=1.05, max_translation=0.05):
    # Apply random rotation (within 10 degrees total range: ±5 degrees)
    rotation_matrix = random_rotation_matrix(max_angle_degrees)
    
    # Apply random scaling (0.95 to 1.05)
    scaling_matrix = random_scaling_matrix(min_scale, max_scale)
    
    # Apply very little translation
    translation_matrix = random_translation_matrix(max_translation)
    
    # Combine all transformations
    combined_transform = np.dot(np.dot(rotation_matrix, scaling_matrix), translation_matrix)
    mesh.apply_transform(combined_transform)
    return mesh
```

```python
# Create output directory if it doesn't exist
os.makedirs(output_dir, exist_ok=True)

for filename in os.listdir(source_dir):
    if filename.endswith(".obj"):
        mesh = trimesh.load(os.path.join(source_dir, filename))
        
        # Generate K=5 augmented versions for each original shape
        for i in range(5):
            transformed_mesh = apply_random_transformations(
                mesh.copy(), 
                max_angle_degrees=10,  # rotation within 10 degrees
                min_scale=0.95,        # scaling from 0.95 to 1.05
                max_scale=1.05,
                max_translation=0.05   # very little translation
            )
            output_filename = f"{os.path.splitext(filename)[0]}_transformed_{i}.obj"
            transformed_mesh.export(os.path.join(output_dir, output_filename))
            
        print(f"Processed {filename} - created 5 augmented versions")
```
